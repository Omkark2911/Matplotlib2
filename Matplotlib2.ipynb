### Ploting

```python
#plt.plot(x,y,color='color_name', linestyle='line_style',linewidth=value,maker='maker symbol,label='label_name')
import matplotlib.pyplot as plt

months = ['Jan','Feb','Mar','Apr','May']
sales = [1000,4000,2000,6000,5000]

plt.plot(months,sales,color = 'lightblue',linestyle= '-',linewidth= 1,marker = 'o',label = '2025 sales data')
plt.title('Sales per month')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.legend()
plt.grid(color = 'gray',linestyle=':',linewidth=1)
plt.show()
```

### Bar chart

```python
#if we want to plot it horizintal ......plt.barh()
import matplotlib.pyplot as plt

fruits = ['Apple','Orange','Banana','Kiwi']
sales = [250,400,600,500]

plt.bar(fruits,sales,color = 'orange',linestyle = '--',linewidth = 2,label = 'fruits sales data')
plt.legend()
plt.title('Sales per fruit')
plt.xlabel('Fruits')
plt.ylabel('Sales')
plt.show()
```

### Pie Chart

```python
#plt.pie(values,label=label_list,colors=color_list,autopct='%1.1f%%)
import matplotlib.pyplot as plt

Cuisine = ['Indian', 'Chinese', 'American', 'Japanese']
Hotels = [5000, 2000, 3500, 4000]

plt.pie(Hotels, labels=Cuisine, autopct='%1.1f%%')
plt.title('No. of Hotels')
plt.show()

```

### Histogram

```python
#plt.hist(data, bins=number_of_bins,color='color_name',edgecolor='black)
import matplotlib.pyplot as plt

scores = [21,44,56,45,96,78,65,79,34,57,23,10,6,7,99]

plt.hist(scores,bins = 5,color='pink',edgecolor='black')
plt.title('Test Scores')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.show()
```

### Scatter

```python
#plt.scatter(x,y,color='color_name,marker='marker_style',label = 'label_name')
import matplotlib.pyplot as plt

hours_studies = [1,2,3,4,5,6,7,8]
exam_scores = [50,55,65,70,80,85,90,95]

plt.scatter(hours_studies,exam_scores,color='gray',marker='s',label='student data')
plt.xlabel('Hours of study')
plt.ylabel('Exam Scores')
plt.title('Study Hours vs Exam Scores')
plt.legend()
plt.grid()
plt.show()

```

## Subplot functions

```python
#plt.subplot(nrows,ncolumns,index)
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,35,30,25]

plt.subplot(1,2,1)
plt.plot(x,y,marker='s')
plt.title('Line chart')

plt.subplot(1,2,2)
plt.bar(x,y)
plt.title('Bar chart')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
#fig, ax = plt.subplot(nrows,ncolumn,figuresize=(width,height))
fig, ax = plt.subplots(1,2,figsize=(12,6))
x = [1,2,3,4]
y =[10,20,15,25]

ax[0].plot(x,y)
ax[0].set_title('Line Chart')

ax[1].bar(x,y)
ax[1].set_title('Bar Chart')

plt.tight_layout
plt.show()
```

## Saving figures to .png, .jpg, .pdf, etc

```python
#savefig('filename,extension',dpi = value,bbox_inches = 'tight')
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,35,30,25]

plt.plot(x,y,color='purple',marker='+')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line chart')

plt.savefig('line_chart.png',dpi = 300,bbox_inches='tight')
plt.show()
```
